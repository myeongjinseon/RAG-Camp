# Chain

## Lambda?

Anonymous function

### Normally

```python
def increment(x):
    return x + 1

result = increment(2)
print(result)
```

### Using Lambda

```python
increment = lambda x: x + 1

result = increment(2)
print(result)
```

### RunnableLambda

```python
from langchain_core.runnables import RunnableLambda
```

```python
increment = RunnableLambda(lambda x: x + 1)

result = increment.invoke(2)
print(result)
```

## Load Environment Variables

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
# retrieve API key and other environment variables
load_dotenv()
```

## Model and Prompt

```python
model = ChatOpenAI(
    model_name="gpt-6-luna",
)
```

```python
prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system", "You are an expert translator who translates {input_language} to {output_language}.",
        ),
        (
            "human", "{text}"
        )
    ]
)
```

## Base

```python
chain = prompt | model
```

```python
out = chain.invoke(
    {
        "input_language": "English",
        "output_language": "Korean",
        "text": "For God so loved the world, that he gave his only begotten Son, that whosoever believeth in him should not perish, but have everlasting life.",
    }
)
```

```python
out
```

```python
out.content
```

## StrOutputParser

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
chain = prompt | model | StrOutputParser()
```

```python
out = chain.invoke(
    {
        "input_language": "English",
        "output_language": "Korean",
        "text": "For God so loved the world, that he gave his only begotten Son, that whosoever believeth in him should not perish, but have everlasting life.",
    }
)
```

```python
out
```

## Split Text

```python
split_text = RunnableLambda(lambda text: text.split())
```

```python
chain = prompt | model | StrOutputParser() | split_text
```

```python
out = chain.invoke(
    {
        "input_language": "English",
        "output_language": "Korean",
        "text": "For God so loved the world, that he gave his only begotten Son, that whosoever believeth in him should not perish, but have everlasting life.",
    }
)
```

```python
out
```

## Count Words

```python
count_words = RunnableLambda(lambda word_list: len(word_list))
```

```python
# one-line
chain = prompt | model | StrOutputParser() | split_text | count_words
```

```python
# multi-lines
chain = (
    prompt
    | model
    | StrOutputParser()
    | split_text
    | count_words
)
```

```python
out = chain.invoke(
    {
        "input_language": "English",
        "output_language": "Korean",
        "text": "For God so loved the world, that he gave his only begotten Son, that whosoever believeth in him should not perish, but have everlasting life.",
    }
)
```

```python
out
```
