# Template

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
# retrieve API key and other environment variables
load_dotenv()
```

```python
model = ChatOpenAI(
    model_name="gpt-6-luna",
)
```

## Define Template

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt = PromptTemplate.from_template(
    """
    당신은 세계에서 매우 유명한 {kind} 요리사입니다.
    아래 제가 가지고 있는 재료를 가지고 레시피 한 개를 추천해주세요.
    요리 이름을 말하고, 사용된 재료를 나열하고, 마지막으로 조리 방법을 설명해주세요.
    <재료>
    {ingredients}
    """
)
```

```python
prompt
```

## 이태리식

```python
prompt_msg = prompt.format(kind="이태리", ingredients="버섯, 파슬리, 면, 피망, 오이, 양파")
```

```python
print(prompt_msg)
```

```python
for chunk in model.stream(prompt_msg):
    print(chunk.content, end="", flush=True)
```

## 한식

```python
prompt_msg = prompt.format(kind="한식", ingredients="소고기, 당근, 양파, 시금치")
```

```python
print(prompt_msg)
```

```python
for chunk in model.stream(prompt_msg):
    print(chunk.content, end="", flush=True)
```
