# Temperature

⚠️ The latest GPT models such as `gpt-6-luna` do not support the temperature option (because of its reasoning capability).

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
# retrieve API key and other environment variables
load_dotenv()
```

```python
# helper function for displaying markdown
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
question = "빅뱅에 대해서 설명해주세요."
```

## temperature=0

```python
chat1 = ChatOpenAI(
    model_name="gpt-4o-mini",
    temperature=0
)
```

### First Try

```python
out = chat1.invoke(question)
```

```python
printmd(out.content)
```

### Second Try

```python
out = chat1.invoke(question)
```

```python
printmd(out.content)
```

## temperature=1

```python
chat2 = ChatOpenAI(
    model_name="gpt-4o-mini",
    temperature=1
)
```

### First Try

```python
out = chat2.invoke(question)
```

```python
printmd(out.content)
```

### Second Try

```python
out = chat2.invoke(question)
```

```python
printmd(out.content)
```
