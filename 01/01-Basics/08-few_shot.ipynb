# Few Shot

⚠️ The latest GPT models know how to compose 삼행시.

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
# retrieve API key and other environment variables
load_dotenv()
```

```python
# helper function for displaying markdown
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
model = ChatOpenAI(
    model_name="gpt-3.5-turbo",
    temperature=0
)
```

## Default

```python
query = "손흥민으로 삼행시 만들어줘."
```

```python
out = model.invoke(query)
```

```python
printmd(out.content)
```

## Few-Shot

### Set Up Few Shot Template

```python
from langchain_core.prompts import FewShotPromptTemplate, PromptTemplate
```

```python
examples = [
    {
        "question": "홍길동으로 삼행시 만들어줘",
        "answer":
            """
            홍: 홍수가 나도
            길: 길이 아무리 험해도
            동: 동에 번쩍 서에 번쩍
            """
    }
]        
```

```python
example_prompt = PromptTemplate(
    input_variables=["question", "answer"],
    template="Question: {question}\n{answer}"
)
```

```python
print(example_prompt.format(**examples[0]))
```

```python
prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    suffix="Question: {input}",
    input_variables=["input"]
)
```

### Example 1

```python
out = model.invoke(prompt.format(input=query))
```

```python
print(out.content)
```

### Example 2

```python
out = model.invoke(prompt.format(input="호날두로 삼행시 만들어줘"))
```

```python
print(out.content)
```

### Example 3

```python
out = model.invoke(prompt.format(input="전재영으로 삼행시 만들어줘"))
```

```python
print(out.content)
```
