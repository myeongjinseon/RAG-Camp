# Output Parser

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
```

```python
# retrieve API key and other environment variables
load_dotenv()
```

```python
# helper function for displaying markdown
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
model = ChatOpenAI(
    model_name="gpt-6-luna",
)
```

## Basic

```python
prompt1 = PromptTemplate(
    template="Give me {how_many} famous K-Pop artists.",
    input_variables=["how_many"]
)
```

```python
prompt1
```

```python
chain1 = prompt1 | model
```

```python
out1 = chain1.invoke(
    {
        "how_many": 3
    }
)
```

```python
out1
```

```python
type(out1)
```

```python
printmd(out1.content)
```

```python
print(out1.content)
```

```python
type(out1.content)
```

## Manual Parsing

```python
import re
```

```python
# use regular expression to extract only the artist names
matches = re.findall(r"^-\s(.*)", out1.content, re.MULTILINE)
```

```python
matches
```

```python
type(matches)
```

## Output Parser 

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
output_parser = CommaSeparatedListOutputParser()
format_instructions = output_parser.get_format_instructions()
```

```python
format_instructions
```

```python
prompt2 = PromptTemplate(
    template="Give me {how_many} famous K-Pop artists.\n{format_instructions}",
    input_variables=["how_many"],
    partial_variables={"format_instructions": format_instructions}
)
```

```python
prompt2
```

```python
chain2 = prompt2 | model | output_parser
```

```python
out2 = chain2.invoke(
    {
        "how_many": 3
    }
)
```

```python
out2
```

```python
type(out2)
```
