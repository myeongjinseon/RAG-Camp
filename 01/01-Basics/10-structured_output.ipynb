```python
model = ChatOpenAI(
    model_name="gpt-6-luna"
)
```

## Normally

## With Structured Output

### Schema

Pydantic is a powerful data validation library.

```python
from pydantic import BaseModel, Field
```

```python
class ResponseFormatter(BaseModel):
    result: str = Field(description="'Yes' or 'No'")
```

```python
model_with_structure = model.with_structured_output(ResponseFormatter)
```

### Example 1

```python
response = model_with_structure.invoke("Is Albert Einstein Jewish?")
```

```python
response
```

```python
response.result
```

### Example 2

```python
response = model_with_structure.invoke("Is Albert Einstein Korean?")
```

```python
response
```

```python
response.result
```
