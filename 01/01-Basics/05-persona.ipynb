# Persona

```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
# retrieve API key and other environment variables
load_dotenv()
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import SystemMessage
from langchain_core.prompts import HumanMessagePromptTemplate
```

```python
# helper function for displaying markdown
from IPython.display import Markdown, display
def printmd(string):
    display(Markdown(string))
```

```python
question1 = "빅뱅에 대해서 설명해주세요."
question2 = "인공지능에 대해서 설명해주세요."
```

```python
model = ChatOpenAI(
    model_name="gpt-6-luna",
)
```

## Professor

```python
chat_template1 = ChatPromptTemplate.from_messages(
    [
        SystemMessage(
            content="당신은 물리학과 대학교 교수입니다. 당신의 청중은 대학생입니다."
        ),
        HumanMessagePromptTemplate.from_template("{text}")
    ]
)
```

### Question 1

```python
messages1 = chat_template1.format_messages(text=question1)
```

```python
messages1
```

```python
%%time

out1 = model.invoke(messages1)
```

```python
printmd(out1.content)
```

### Question 2

```python
messages2 = chat_template1.format_messages(text=question2)
```

```python
messages2
```

```python
%%time

out2 = model.invoke(messages2)
```

```python
printmd(out2.content)
```
