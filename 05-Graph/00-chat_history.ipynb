## With History

### Construct History

```python
from langchain_core.messages import AIMessage, HumanMessage
```

```python
msg1 = [
    AIMessage(
        content="당신의 이름은 무엇입니까?",
        name="Model"
    )
]
```

```python
msg2 = [
    HumanMessage(
        content="내 이름은 재영이야.",
        name="User"
    )
]
```

```python
msg3 = [
    AIMessage(
        content="몇 살이신지 물어봐도 될까요?",
        name="Model"
    )
]
```

```python
msg4 = [
    HumanMessage(
        content="그럼! 나는 12살이야.",
        name="User"
    )
]
```

```python
msg5 = [
    AIMessage(
        content="오늘 무엇을 도와드릴까요?",
        name="Model"
    )
]
```

```python
msg6 = [
    HumanMessage(
        content="내 이름을 말해줘. 그리고 내가 이해할 수 있는 수준으로 LLM에 대해 설명해줄 수 있어?",
        name="User"
    )
]
```

```python
messages = []
```

```python
messages.extend(msg1)
messages.extend(msg2)
messages.extend(msg3)
messages.extend(msg4)
messages.extend(msg5)
messages.extend(msg6)
```

```python
for msg in messages:
    msg.pretty_print()
```

### Invoke

```python
result = model.invoke(messages)
```

```python
result.pretty_print()
```
