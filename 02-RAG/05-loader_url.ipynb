## Crawling

```python
from langchain_community.document_loaders import WebBaseLoader
```

```python
# 세계의 홍수 전설
# url = "https://creation.kr/EvidenceofFlood/?idx=1288372&bmode=view"
url = "https://www.chpress.net/column-detail.html?cate=column&c_id=84&id=8503"
```

### Entire Page

```python
loader = WebBaseLoader(web_path=url)
```

```python
page = loader.load()
```

```python
page
```

```python
print(page[0].page_content)
```

### Only Content

```python
import bs4
import re
```

```python
# using BeautifulSoup
loader = WebBaseLoader(
    web_path=url,
    bs_kwargs=dict(
        parse_only=bs4.SoupStrainer(
            class_=("col-lg-8 col-md-8 col-sm-12 col-xs-12")
        )
    ),
)

page = loader.load()
```

```python
# using regular expression
match = re.search(r"창조과학 칼럼(.*?)▲이메일:", page[0].page_content, re.DOTALL)
if match:
    page[0].page_content = match.group(1)
```

```python
page
```

## Splitting

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)
```

```python
texts = text_splitter.split_documents(page)
```

```python
texts
```

```python
len(texts)
```

## Embedding

```python
from langchain_openai import OpenAIEmbeddings
```

```python
embedding_model = OpenAIEmbeddings(model="text-embedding-3-small")
```

```python
text_content = [text.page_content for text in texts]
```

```python
text_content
```

```python
embeddings = embedding_model.embed_documents(text_content)
```
